# 02 · Retrieve company information

**LSEG Workspace · CodeBook · Python**

**Objective.** Retrieve a corporate profile for one company, then repeat the same request for multiple companies. Use the precise instrument RIC and verify the underlying company entity in Workspace.

The `fields` list may be adapted in CodeCreator according to the organization's coverage and the available data items.

## Step 1 · Open CodeBook session and select a company

```python
import refinitiv.data as rd
import pandas as pd
from pathlib import Path

rd.open_session()
company_ric = "IBE.MC"   # Iberdrola: example instrument RIC
```

## Step 2 · Select the company profile fields

```python
fields = [
    "TR.CommonName",             # Company name
    "TR.OrganizationID",         # Organization PermID
    "TR.HeadquartersCountry",    # Headquarters country
    "TR.TRBCEconomicSector",     # Economic sector
    "TR.TRBCBusinessSector",     # Business sector
    "TR.TRBCIndustry",           # Industry
    "TR.BusinessSummary",        # Main business activities
]
```

## Step 3 · Retrieve the company profile

```python
company_info = rd.get_data(universe=company_ric, fields=fields)
display(company_info.T)
```

## Step 4 · Retrieve the same information for several companies

```python
company_rics = [
    "IBE.MC",   # Iberdrola
    "SAN.MC",   # Banco Santander
    "ITX.MC",   # Inditex
]

company_profiles = rd.get_data(universe=company_rics, fields=fields)
company_profiles = (
    company_profiles.sort_values(by="Company Common Name")
    .reset_index(drop=True)
)
display(company_profiles)
```

## Step 5 · Export the results

```python
output_folder = Path("output")
output_folder.mkdir(parents=True, exist_ok=True)
company_profiles.to_csv(
    output_folder / "company_profiles_spain.csv",
    index=False,
    encoding="utf-8-sig",
)
print("Saved company_profiles_spain.csv")
```

**Practice.** Replace one RIC with another verified in Workspace. A RIC identifies an instrument; the organization PermID identifies an entity. Do not treat them as interchangeable.
