**LSEG Workspace · CodeBook · Python**

**Objective.** Download the latest reported annual financial statement figures for three firms, then use the same companies to retrieve a monthly *market-price* series for a specific period. Monthly prices are **not** monthly financial statements.

The accounting line items below are illustrative: confirm exact field codes, reporting basis, currency conversion and scale in CodeCreator. Banks and non-financial firms can have different reporting structures; do not compare their financial metrics mechanically.

## Step 1 · Import libraries and select companies

```python
import refinitiv.data as rd
import pandas as pd
from pathlib import Path

rd.open_session()
company_rics = ["IBE.MC", "SAN.MC", "ITX.MC"]
output_folder = Path("output")
output_folder.mkdir(parents=True, exist_ok=True)
```

## Step 2 · Download latest reported annual financial data

```python
# Latest reported fiscal year: FY0. Monetary fields: requested in EUR.
financial_fields = [
    "TR.CommonName",
    "TR.F.PeriodEndDate",
    "TR.F.TotAssets",
    "TR.F.TotRevenue",
    "TR.F.NetIncAfterTax",
    "TR.F.DebtTot",
    "TR.F.TotLiab",
]

financial_data = rd.get_data(
    universe=company_rics,
    fields=financial_fields,
    parameters={"Period": "FY0", "Curn": "EUR", "Scale": 6},
)
display(financial_data)

financial_data.to_csv(
    output_folder / "financial_data_spain.csv",
    index=False,
    encoding="utf-8-sig",
)
```

## Step 3 · Download monthly market prices for a defined period

```python
# Monthly price observations from January 2024 through December 2025.
# SDate = start date; EDate = end date; Frq = M (monthly).
monthly_prices = rd.get_data(
    universe=company_rics,
    fields=["TR.PriceClose.Date", "TR.PriceClose"],
    parameters={
        "SDate": "2024-01-01",
        "EDate": "2025-12-31",
        "Frq": "M",
    },
)
display(monthly_prices)

monthly_prices.to_csv(
    output_folder / "monthly_prices_spain_2024_2025.csv",
    index=False,
    encoding="utf-8-sig",
)
print("Financial and monthly-price CSV files saved in:", output_folder.resolve())
```

**Practice.** Change the dates and frequency (`"Frq": "D"` for daily observations), keeping the same instruments. Observe that fiscal periods such as `FY0` specify *financial reporting periods*, whereas `SDate`, `EDate` and `Frq` specify *market-data observation dates and frequency*.
