## Connect and select companies

```python
import refinitiv.data as rd
import pandas as pd
from pathlib import Path

rd.open_session()
company_rics = ["IBE.MC", "SAN.MC", "ITX.MC"]
```

## Request annual fundamentals

```python
# FY0 = reported fiscal-year values; Frq=FY requests annual observations.
# Period-end dates can differ across companies and calendar years.
fields = [
    "TR.CommonName", "TR.F.TotAssets.periodenddate",
    "TR.F.TotAssets", "TR.F.TotRevenue", "TR.F.NetIncAfterTax"
]
financial_history = rd.get_data(
    universe=company_rics,
    fields=fields,
    parameters={"SDate":"2021-01-01", "EDate":"2025-12-31",
                "Frq":"FY", "Period":"FY0", "Curn":"EUR", "Scale":6}
)
display(financial_history.head(15))
```

## Export

```python
output_folder = Path("output")
output_folder.mkdir(parents=True, exist_ok=True)
output_file = output_folder / "04_historical_financial_data.csv"
financial_history.to_csv(output_file, index=False, encoding="utf-8-sig")
print("Saved:", output_file.resolve())
```
