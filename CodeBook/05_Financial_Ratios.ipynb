## Connect and select companies

```python
import refinitiv.data as rd
import pandas as pd
from pathlib import Path

rd.open_session()
company_rics = ["IBE.MC", "SAN.MC", "ITX.MC"]
```

## Request ratios

```python
# Ratio definitions and percentage units must be checked in CodeCreator.
fields = [
    "TR.CommonName",
    "TR.F.ReturnAvgTotAssetsPct",
    "TR.F.ReturnAvgComEqPct",
    "TR.F.TotDebtPctofTotEq"
]
ratios = rd.get_data(
    universe=company_rics, fields=fields,
    parameters={"Period":"FY0"}
)
display(ratios)
```

## Export

```python
output_folder = Path("output")
output_folder.mkdir(parents=True, exist_ok=True)
output_file = output_folder / "05_financial_ratios.csv"
ratios.to_csv(output_file, index=False, encoding="utf-8-sig")
print("Saved:", output_file.resolve())
```
