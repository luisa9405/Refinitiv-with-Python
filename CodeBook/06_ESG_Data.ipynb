## Connect and select companies

```python
import refinitiv.data as rd
import pandas as pd
from pathlib import Path

rd.open_session()
company_rics = ["IBE.MC", "SAN.MC", "ITX.MC"]
```

## Request ESG scores

```python
fields = [
    "TR.CommonName", "TR.TRESGScore", "TR.TRESGCScore",
    "TR.EnvironmentPillarScore", "TR.SocialPillarScore",
    "TR.GovernancePillarScore"
]
esg_data = rd.get_data(universe=company_rics, fields=fields)
display(esg_data)
```

## Export

```python
output_folder = Path("output")
output_folder.mkdir(parents=True, exist_ok=True)
output_file = output_folder / "06_esg_scores.csv"
esg_data.to_csv(output_file, index=False, encoding="utf-8-sig")
print("Saved:", output_file.resolve())
```
