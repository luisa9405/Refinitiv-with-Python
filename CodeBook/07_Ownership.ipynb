## Connect

```python
import refinitiv.data as rd
import pandas as pd
from pathlib import Path

rd.open_session()
company_rics = ["IBE.MC", "SAN.MC", "ITX.MC"]
```

## Request ownership statistics

```python
# StatType=7 is an investor-category ownership breakdown, not a list of named holders.
# SDate sets the requested snapshot date; the field may return several categories.
ownership = rd.get_data(
    universe=["IBE.MC"],
    fields=["TR.CategoryOwnershipPct", "TR.InstrStatTypeValue"],
    parameters={"StatType":7, "SDate":"2025-12-31"}
)
display(ownership)
```

## Export

```python
output_folder = Path("output")
output_folder.mkdir(parents=True, exist_ok=True)
output_file = output_folder / "07_ownership_categories.csv"
ownership.to_csv(output_file, index=False, encoding="utf-8-sig")
print("Saved:", output_file.resolve())
```
