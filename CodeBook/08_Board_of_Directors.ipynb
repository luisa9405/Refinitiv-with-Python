## Connect and select companies

```python
import refinitiv.data as rd
import pandas as pd
from pathlib import Path

rd.open_session()
company_rics = ["IBE.MC", "SAN.MC", "ITX.MC"]
```

## Request board indicators

```python
# Check whether the two percentage fields return 0–100 units in your account.
fields = [
    "TR.CommonName", "TR.BoardSize",
    "TR.AnalyticBoardFemale", "TR.AnalyticIndepBoard"
]
board_data = rd.get_data(universe=company_rics, fields=fields)
display(board_data)
```

## Export

```python
output_folder = Path("output")
output_folder.mkdir(parents=True, exist_ok=True)
output_file = output_folder / "08_board_indicators.csv"
board_data.to_csv(output_file, index=False, encoding="utf-8-sig")
print("Saved:", output_file.resolve())
```
