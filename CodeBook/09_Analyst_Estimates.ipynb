## Connect and select companies

```python
import refinitiv.data as rd
import pandas as pd
from pathlib import Path

rd.open_session()
company_rics = ["IBE.MC", "SAN.MC", "ITX.MC"]
```

## Request estimates

```python
# FY1 is a forward fiscal period, not the last reported year.
fields = ["TR.CommonName", "TR.EPSMean", "TR.RevenueMean"]
estimates = rd.get_data(
    universe=company_rics, fields=fields,
    parameters={"Period":"FY1", "Curn":"EUR", "Scale":6}
)
display(estimates)
```

## Export

```python
output_folder = Path("output")
output_folder.mkdir(parents=True, exist_ok=True)
output_file = output_folder / "09_analyst_estimates.csv"
estimates.to_csv(output_file, index=False, encoding="utf-8-sig")
print("Saved:", output_file.resolve())
```
