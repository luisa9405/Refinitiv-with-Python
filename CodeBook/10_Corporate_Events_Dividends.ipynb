## Connect

```python
import refinitiv.data as rd
import pandas as pd
from pathlib import Path

rd.open_session()
company_rics = ["IBE.MC", "SAN.MC", "ITX.MC"]
```

## Retrieve dividends

```python
# DateType=ED applies the date filter to the ex-dividend date.
dividends = rd.get_data(
    universe=["IBE.MC", "ITX.MC"],
    fields=["TR.DivExDate", "TR.DivPayDate", "TR.DivUnadjustedGross", "TR.DivCurr", "TR.DivType"],
    parameters={"SDate":"2023-01-01", "EDate":"2025-12-31", "DateType":"ED"}
)
display(dividends.head(20))
```

## Export

```python
output_folder = Path("output")
output_folder.mkdir(parents=True, exist_ok=True)
output_file = output_folder / "10_dividend_events.csv"
dividends.to_csv(output_file, index=False, encoding="utf-8-sig")
print("Saved:", output_file.resolve())
```
