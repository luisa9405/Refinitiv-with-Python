## Connect

```python
import refinitiv.data as rd
import pandas as pd
from pathlib import Path
rd.open_session()
```

## Expand an index chain

```python
# A chain RIC represents the constituent list; it is not the index-level price RIC.
index_chain = "0#.STOXX50E"
constituents = rd.get_data(
    universe=index_chain,
    fields=["TR.CommonName", "TR.RIC"]
)
display(constituents.head(20))
print("Constituent rows returned:", len(constituents))
```

## Export

```python
output_folder = Path("output")
output_folder.mkdir(parents=True, exist_ok=True)
output_file = output_folder / "11_index_constituents.csv"
constituents.to_csv(output_file, index=False, encoding="utf-8-sig")
print("Saved:", output_file.resolve())
```
