## Connect and select companies

```python
import refinitiv.data as rd
import pandas as pd
from pathlib import Path

rd.open_session()
company_rics = ["IBE.MC", "SAN.MC", "ITX.MC"]
```

## Download a comparison table

```python
fields = ["TR.CommonName", "TR.TRBCEconomicSector", "TR.F.TotAssets", "TR.F.NetIncAfterTax"]
peers = rd.get_data(
    universe=company_rics, fields=fields,
    parameters={"Period":"FY0", "Curn":"EUR", "Scale":6}
)
display(peers)
```

## Export

```python
output_folder = Path("output")
output_folder.mkdir(parents=True, exist_ok=True)
output_file = output_folder / "12_peer_comparison.csv"
peers.to_csv(output_file, index=False, encoding="utf-8-sig")
print("Saved:", output_file.resolve())
```
