## Connect

```python
import refinitiv.data as rd
import pandas as pd
from pathlib import Path
rd.open_session()
```

## Download monthly FX observations

```python
# EUR= and GBP= are market FX identifiers; BID is a quote, not a GDP or CPI value.
fx_data = rd.get_history(
    universe=["EUR=", "GBP="],
    fields=["BID"],
    start="2024-01-01", end="2025-12-31",
    interval="monthly"
)
display(fx_data.head(12))
```

## Export

```python
# get_history typically returns a date-indexed DataFrame; retain dates in the CSV.
output_folder = Path("output")
output_folder.mkdir(parents=True, exist_ok=True)
output_file = output_folder / "13_monthly_fx_quotes.csv"
fx_data.to_csv(output_file, index=True, encoding="utf-8-sig")
print("Saved:", output_file.resolve())
```
