## Connect and select companies

```python
import refinitiv.data as rd
import pandas as pd
from pathlib import Path

rd.open_session()
company_rics = ["IBE.MC", "SAN.MC", "ITX.MC"]
```

## Request firm-year fundamentals

```python
# The period-end date (rather than the extraction year) identifies each record.
fields = [
    "TR.CommonName", "TR.F.TotAssets.periodenddate",
    "TR.F.TotAssets", "TR.F.TotRevenue"
]
panel = rd.get_data(
    universe=company_rics, fields=fields,
    parameters={"SDate":"2022-01-01", "EDate":"2025-12-31",
                "Period":"FY0", "Frq":"FY", "Curn":"EUR", "Scale":6}
)
# Inspect the returned date field: its displayed label may vary by library version.
display(panel.head(20))
```

## Export

```python
output_folder = Path("output")
output_folder.mkdir(parents=True, exist_ok=True)
output_file = output_folder / "14_simple_firm_year_panel.csv"
panel.to_csv(output_file, index=False, encoding="utf-8-sig")
print("Saved:", output_file.resolve())
```
